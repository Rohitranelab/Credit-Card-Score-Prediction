```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r'../data/credit_score_prediction.csv')
df.head()
```

```python
df.info()
```

```python
df.drop(columns = ['customer_id'], axis = 1, inplace = True)
```

```python
df.head()
```

```python
df['payment_history'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df['payment_history'] = le.fit_transform(df['payment_history'])
```

```python
df.head()
```

```python
df['num_credit_cards'].value_counts()
```

```python
x = df.drop(columns = ['credit_score'], axis = 1)
y = df['credit_score']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
x_train.shape, x_test.shape
```

```python
y_train.shape, y_test.shape
```

```python
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, AdaBoostRegressor, GradientBoostingRegressor, ExtraTreesRegressor
from xgboost import XGBRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, root_mean_squared_error
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(),
    "Lasso Regression": Lasso(),
    "Decision Tree": DecisionTreeRegressor(random_state = 42),
    "Random Forest": RandomForestRegressor(n_estimators = 100, random_state = 42),
    "Gradient Boosting": GradientBoostingRegressor(random_state = 42),
    "AdaBoost": AdaBoostRegressor(random_state = 42),
    "Support Vector Regressor": SVR(),
    "K-Nearest Neighbors": KNeighborsRegressor(n_neighbors = 10),
    "XGBoost" : XGBRegressor(n_estimators = 100)
}
```

```python
results = []

for name, model in models.items():
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)

    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = root_mean_squared_error(y_test, y_pred)

    results.append({
        'Model Name: ' : name,
        'R2 Score' : r2,
        'Mean Absolute Error' : mae,
        'Mean Squared Error' : mse,
        'Root Mean Squared Error' : rmse
    })

    print('\n' + '-' * 45)
    print('Model:', name)
    print('R2 Score:', r2)
    print('Mean Absolute Error:', mae)
    print('Mean Squared Error:', mse)
    print('Root Mean Squared Error:', rmse)

summary_df = pd.DataFrame(results).sort_values(by = "R2 Score", ascending = False)
summary_df.index = range(1, len(summary_df) + 1)

print('\n' + '=' * 45)
print('MODEL SUMMARY')
print('=' * 45)
summary_df
```
